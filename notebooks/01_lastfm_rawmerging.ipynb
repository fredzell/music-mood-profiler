# Last.fm Raw Data Merging

This notebook merges yearly Last.fm exports into a single dataset and performs an initial inspection of the collected listening history.

## 1. Load libraries

```python
import pandas as pd
from pathlib import Path
```

## 2. Load the data

```python
files = sorted(Path("../data/raw/lastfm").glob("scrobbles_*.parquet"))
print(files)
```

## 3. Combine the data into a single dataframe

```python
dfs = []

for f in files:
    year = f.stem.split("_")[-1]
    
    if not year.isdigit():
        print(f"Skipping file {f}.")
        continue
    
    year = int(year)
    temp = pd.read_parquet(f)
    temp["year_file"] = year
        
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)
```

```python
df
```

## 4. Export merged dataset

```python
df.to_parquet("../data/interim/lastfm/lastfm_scrobbles_merged.parquet", index=False)
```
