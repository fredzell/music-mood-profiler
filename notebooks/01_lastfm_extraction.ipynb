```python
import pandas as pd
from pathlib import Path
```

```python
files = sorted(Path("../data/raw/lastfm").glob("scrobbles_*.parquet"))
print(files)
```

```python
dfs = []

for f in files:
    year = f.stem.split("_")[-1]
    
    if not year.isdigit():
        print(f"Skipping file {f}.")
        continue
    
    year = int(year)
    temp = pd.read_parquet(f)
    temp["year_file"] = year
        
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)
```
